# Conditional Statements

## 28. Determine the generation and birth year for `age = 26`

Using the generation ranges from The College Investor, Gen Z covers birth years 1997 through 2012
and Millennials cover birth years 1981 through 1996. The birth year is calculated from the current
year so the logic stays correct over time.

```python
from datetime import date

age = 26
current_year = date.today().year
birth_year = current_year - age

if 1997 <= birth_year <= 2012:
    generation = "Gen Z"
elif 1981 <= birth_year <= 1996:
    generation = "a Millennial"
else:
    generation = "neither Gen Z nor a Millennial"

print(f"A person who is {age} years old was born in {birth_year}, which makes them {generation}.")
```

## 29. Temperature conversion function

The function accepts a list of numbers and an optional scale argument that defaults to `"F"`.
Fahrenheit values are printed as given, while `"C"` and `"K"` are converted and rounded to two
decimal places. Any other letter produces an error message.

```python
def print_temperatures(temperatures, scale="F"):
    """Print each temperature in the requested scale.

    temperatures : list of numbers, assumed to be in degrees Fahrenheit
    scale        : "F" for Fahrenheit (default), "C" for Celsius, "K" for Kelvin
    """
    scale = scale.upper()

    if scale == "F":
        for temp in temperatures:
            print(f"The temperature is {temp} degrees Fahrenheit")
    elif scale == "C":
        for temp in temperatures:
            celsius = round((temp - 32) * 5 / 9, 2)
            print(f"The temperature is {celsius} degrees Celsius")
    elif scale == "K":
        for temp in temperatures:
            kelvin = round((temp - 32) * 5 / 9 + 273.15, 2)
            print(f"The temperature is {kelvin} degrees Kelvin")
    else:
        print(f"Error: '{scale}' is not a valid scale. Please use 'F', 'C', or 'K'.")


temperature_list = [32, 55, 90, 212]

# Default behavior, no second argument supplied
print_temperatures(temperature_list)
```

## 30. Call the function in degrees Celsius

```python
print_temperatures(temperature_list, "C")
```

## 31. Call the function in degrees Kelvin

```python
print_temperatures(temperature_list, "K")
```

## Bonus check: an invalid scale triggers the error message

```python
print_temperatures(temperature_list, "X")
```
